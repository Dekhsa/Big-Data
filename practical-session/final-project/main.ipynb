# DATA PREPROCESSING


```python
message = df['message']
message
```

```python
# convert message to lowercase
message = message.str.lower()
message
```

```python
string.punctuation
```

```python
# Remove punctuation(tanda baca) function
def remove_punctuation(text):
   return text.translate(str.maketrans('', '', string.punctuation))
```

```python
message = message.apply(lambda x: remove_punctuation(x))
message
```
