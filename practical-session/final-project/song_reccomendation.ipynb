## Data Cleaning


### Remove Non-Indonesian/English Words

using langdetect


```python
from langdetect import detect, LangDetectException

def is_indonesian_or_english(text):
    # Only process non-empty strings
    if not isinstance(text, str) or not text.strip():
        return False
    try:
        # Detect the language of the text
        lang = detect(text)
        # Return True if the language is Indonesian or English
        return lang in ['id', 'en']
    except LangDetectException:
        # If language detection fails, assume it's not the desired language
        return False
```

```python
# Apply the function to the 'message' column and keep only the rows that return True
df = df[df['message'].apply(is_indonesian_or_english)]

# Display the first few rows of the filtered DataFrame
df
```

```python
df.to_csv('indo-english-messageDataset.csv', index=False, encoding='utf-8')
```

### Delete Duplicate Data

### Convert Message to Lowercase


```python
df['clean_message'] = df['message'].str.lower()
df
```

### Remove Punctuations


```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
df['clean_message'] = df['clean_message'].apply(lambda x: remove_punctuation(x) if isinstance(x, str) else x)
df[['message', 'clean_message']]
```

### Removal of Stopwords

using NLTK or natural language toolkit


```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

print("Indoensian: ", " ".join(stopwords.words('indonesian')))
print("english   : ", " ".join(stopwords.words('english')))
```

```python
STOPWORDS = set(stopwords.words('indonesian') + stopwords.words('english'))
def remove_stopwords(text):
   return " ".join([word for word in text.split() if word not in STOPWORDS])
```

```python
df['clean_message'] = df['clean_message'].apply(lambda x: remove_stopwords(x) if isinstance(x, str) else x)
df[['message', 'clean_message']]
```

### Removal of Frequent Words


```python
from collections import Counter
word_count = Counter()
for text in df['clean_message']:
    if isinstance(text, str):
        for word in text.split():
             word_count[word] += 1

word_count.most_common(10)
```

```python
FREQUENT_WORDS = set(word for (word, wc) in word_count.most_common(3))
def remove_frequent_words(text):
    return " ".join([word for word in text.split() if word not in FREQUENT_WORDS])
```

```python
df['clean_message'] = df['clean_message'].apply(lambda x: remove_frequent_words(x) if isinstance(x, str) else x)
df[['message', 'clean_message']]
```

### Remove Emojis


```python
import re

def remove_emojis(text):
    # Return the original value if it's not a string
    if not isinstance(text, str):
        return text
    
    # Comprehensive regex to find most emojis
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map symbols
        "\U0001F700-\U0001F77F"  # alchemical symbols
        "\U0001F780-\U0001F7FF"  # Geometric Shapes Extended
        "\U0001F800-\U0001F8FF"  # Supplemental Arrows-C
        "\U0001F900-\U0001F9FF"  # Supplemental Symbols and Pictographs
        "\U0001FA00-\U0001FA6F"  # Chess Symbols
        "\U0001FA70-\U0001FAFF"  # Symbols and Pictographs Extended-A
        "\U00002702-\U000027B0"  # Dingbats
        "\U000024C2-\U0001F251" 
        "]+",
        flags=re.UNICODE,
    )
    # Remove all emojis from the string
    return emoji_pattern.sub(r'', text)

# Apply the function to the 'stemmed_message' column
df['clean_message'] = df['clean_message'].apply(remove_emojis)

# Display the column to verify the changes
df[['message', 'clean_message']]
```

### Remove Rare Words


```python
RARE_WORDS = set(word for (word, wc) in word_count.most_common()[:-10:-1])  # Get the last 10 rarest words
RARE_WORDS
```

```python
def remove_rare_words(text):
   return " ".join([word for word in text.split() if word not in RARE_WORDS])
```

```python
df['clean_message'] = df['clean_message'].apply(lambda x: remove_rare_words(x) if isinstance(x, str) else x)
df[['message', 'clean_message']]
```

### Remove Special Characters


```python
import re
def remove_spl_chars(text):
   text = re.sub('^a-zA-Z0-9', ' ', text)
   text = re.sub('\s+', ' ', text)
   return text
```

```python
df['clean_message'] = df['clean_message'].apply(lambda x: remove_spl_chars(x) if isinstance(x, str) else x)
df[['message', 'clean_message']]
```

## Stemming


```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
def stem_words(text):
   return " ".join([ps.stem(word) for word in text.split()])
```

```python
df['stemmed_message'] = df['clean_message'].apply(lambda x: stem_words(x) if isinstance(x, str) else x)
df[['message', 'clean_message', 'stemmed_message']]
```

## Lemmatization & POS Tagging


```python
import nltk
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
```

```python
from nltk import pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
wordnet_map = {'N': wordnet.NOUN, 'V': wordnet.VERB, 'J': wordnet.ADJ,  'R': wordnet.ADV}

def lemmatize_words(text):
   pos_text = pos_tag(text.split())
   return ' '.join([lemmatizer.lemmatize(word, wordnet_map.get(pos[0], wordnet.NOUN)) for word, pos in pos_text])
```

```python
df['lemmatized_message'] = df['clean_message'].apply(lambda x: stem_words(x) if isinstance(x, str) else x)
df[['message', 'clean_message', 'stemmed_message', 'lemmatized_message']]
```

## Save Cleaned Dataset


```python
cleaned_df = df.drop(columns=['message', 'lemmatized_message', 'stemmed_message'])
cleaned_df
```
